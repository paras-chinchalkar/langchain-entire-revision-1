```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("LangChain_Overview.txt")
docs=loader.load()
```

```python
docs
```

```python
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20)
```

```python
chunk=splitter.split_documents(docs)
```

```python
chunk
```

```python
#2-vector-stores
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(
    model='sentence-transformers/all-MiniLM-L6-v2'
)
```

```python
from langchain_community.vectorstores import FAISS
vector_store=FAISS.from_documents(chunk,embedding=embeddings)
```

```python
vector_store
```

```python
retriever=vector_store.as_retriever(search_type='mmr',search_kwargs={"k":4,"lambda_mult":0.7})
```

```python
retriever
```

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"]=os.getenv('GROQ_API_KEY')
llm=init_chat_model(model="groq:llama-3.1-8b-instant")
llm 
```

```python
#Query decomposition is used to make complex or ambiguous queries easier to handle by breaking them down into smaller, more manageable parts. Think of it like debugging a big piece of code—you split it into modules so each part can be understood and solved clearly.
```

```python
from langchain_core.prompts import PromptTemplate
decomposition_query=PromptTemplate.from_template("""
You are an AI Assistant That will work on the basis of Decomposition prompt,based on Question
                                   
Question:{question}
Sub-question:
""")
```

```python
from langchain_core.output_parsers import StrOutputParser 
decomposition_chain=decomposition_query | llm | StrOutputParser()
```

```python
query="How does langchain work?"
docomposition_question=decomposition_chain.invoke({"question":query})
```
