```python
from langchain_community.document_loaders import TextLoader
```

```python
text_loader=TextLoader("LangChain_Overview.txt")
raw_data=text_loader.load()
```

```python
#Splitter
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20)
```

```python
chunk=splitter.split_documents(raw_data)
```

```python
chunk
```

```python
#2-vector-stores
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(
    model='sentence-transformers/all-MiniLM-L6-v2'
)
```

```python
#Vector-stores

from langchain_community.vectorstores import FAISS
vectorstores=FAISS.from_documents(chunk,embedding=embeddings)
```

```python
vectorstores
```

```python
#MMR retriever
retriever=vectorstores.as_retriever(
    search_type='mmr',
    search_kwargs={"k":3}
)
```

```python
from langchain.chat_models import init_chat_model
llm=init_chat_model(model="groq:llama-3.1-8b-instant")
llm 
```

```python
#Prompt
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser 
query_expansion_prompt=PromptTemplate.from_template(
"""Answer the following question based on the query is imporve docuement retrieval by adding relevant

Query:{query}
Expanded query:
"""
)
#This below is query enhchement with the help of prompt + llm+stroutputparser
query_expansion_chain=query_expansion_prompt| llm | StrOutputParser()
query_expansion_chain
```

```python
#Langchain_query_explansion
query_expansion_chain.invoke({"query":"Langchain Memory"})
```

```python
prompt
```

```python
#Calling the chat model
from langchain.chat_models import init_chat_model
llm=init_chat_model(model="groq:llama-3.1-8b-instant")
llm 
```

```python
llm.invoke("Hi ")
```

```python
#Making an RAG chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_classic.chains import create_retrieval_chain
chain=create_stuff_documents_chain(llm,prompt=prompt)
```

```python
query_expansion_prompt
```

```python
chain
```

```python
#step5:Full TAG Pipeline with query expansion
from langchain_core.runnables import RunnableMap
rag_pipeline=(
    RunnableMap({
        "input":lambda x:x['input'],
        "context":lambda x: retriever.invoke(query_expansion_chain.invoke({"query":x['input']}))

    })
    |create_stuff_documents_chain
)
```
