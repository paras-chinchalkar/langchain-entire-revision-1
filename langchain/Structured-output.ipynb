📘 Pydantic: Structured Explanation
1. Definition
- Pydantic is a Python library for data validation and settings management using Python type hints.
- It ensures that data conforms to the expected types and structures, automatically parsing and validating inputs.




```python
import os
from langchain.chat_models import init_chat_model
os.environ['GROQ_API_KEY']=os.getenv('GROQ_API_KEY')
from langchain_groq import ChatGroq
model=ChatGroq(model="llama-3.1-8b-instant")
```

```python
from pydantic import BaseModel,Field

class Movie(BaseModel):
    title:str=Field(description='The title of the movie')
    year:str=Field(description="This yearmovie was released")
    director:str=Field(description="The director of the movie")
    rating:float=Field(description="The mvoies rating out of 10")
```

```python
model_with_structure=model.with_structured_output(Movie)
```

```python
response=model_with_structure.invoke("Provide details about the movie Inception")
```

```python
response
```

```python
from typing_extensions import TypedDict,Annotated
class MovieDict(TypedDict):
    """A movie with details"""
    title:Annotated[str,...,"The title of the movie is"]
    year:Annotated[str,...,"The year in which the  film was released"]
    director:Annotated[str,...,"The director of the movie is"]
    rating:Annotated[str,...,"The movie's rating out of 10 is"]
```

```python
model_with_typedict=model.with_structured_output(MovieDict)
response=model_with_typedict.invoke("Porvide the info for the movie the shawshank's redemption")
response
```
