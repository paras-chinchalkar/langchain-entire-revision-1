```python
sample_text={
    "data/text_files/Python_intro.txt":"""This module is distinct from langchain_core.messages.content, which provides multimodal content blocks for LLM chat I/O (text, images, audio, etc. within messages).

Key distinction:

Documents (this module): For data retrieval and processing workflows

Vector stores, retrievers, RAG pipelines
Text chunking, embedding, and semantic search
Example: Chunks of a PDF stored in a vector database
Content Blocks (messages.content): For LLM conversational I/O

Multimodal message content sent to/from models
Tool calls, reasoning, citations within chat
Example: An image sent to a vision model in a chat message (via ImageContentBlock)
While both can represent similar data types (text, files), they serve different architectural purposes in LangChain applications."""
}
for filepath,content in sample_text.items():
    with open(filepath,"w",encoding="utf-8") as f:
        f.write(content)
print("The sample file has been created")


```

TextLoader-Read Single File 

```python
from langchain_community.document_loaders import TextLoader

#Loading a single text file
loader=TextLoader("data/text_files/Python_intro.txt",encoding ="utf-8")

document=loader.load()
print(f"Loaded {len(document)} number of documents")
print(f"Loaded {document[0].page_content[:100]}...")
print(f"Metadata:{document[0].metadata}")
```

Now to load multiple text files we will do this step

```python
from langchain_community.document_loaders import DirectoryLoader
dir_loader=DirectoryLoader(
    "data/text_files",
    glob="**/*.txt",##pattern to match fles
    loader_cls=TextLoader ,
    loader_kwargs={'encoding':'utf-8'},
    show_progress=True
)
documents=dir_loader.load()
```

Now in the above step the document is properly loaded

2)Text Splitter strategies

```python
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    TokenTextSplitter
)
print(documents)
```

```python
#Method:Character text spliiter
text=document[0].page_content
text
```

```python
#Method1-character based text splitter
print("Character text splitter")
char_splitter=CharacterTextSplitter(
    separator="\n",
    chunk_size=200,
    chunk_overlap=20,
    length_function=len
    
)
print(char_splitter)
```

```python
char_chunk=char_splitter.split_text(text)
print(f"Created chunks in here are  {len(char_chunk)}")
print(f"Here the first chunk that we have is: {char_chunk[0]}")
```

```python
print(char_chunk[0])
print(char_chunk[1])
```

```python
print("Recursive charcate text splitter here wil be used ")
recursive_splitter=RecursiveCharacterTextSplitter(
    separators=[" "],
    chunk_size=200,
    chunk_overlap=20,
    length_function=len
)
recursive_chunk=recursive_splitter.split_text(text)
print(f"Create chunk lenght is {len(recursive_chunk)}")
# print(f"Fist chunk in here is {recursive_chunk[0]}")
for i in range(len(recursive_chunk)-1):
    print(f"chunk{i+1}:'{recursive_chunk[i]}'")
    print(f"chunk{i+2}:'{recursive_chunk[i]}'")
```

```python
#Text based splitting
print("This is an text based splitting")
token_splitting=TokenTextSplitter(
    chunk_size=50,
    chunk_overlap=10
)
token_chunks=token_splitting.split_text(text)
print(f"the length of the token is {len(token_chunks)}")
print(f"The fIRST CHUNK in here is {token_chunks[0][:100]}")
```
