```python
#Now processing json files
from langchain_community.document_loaders import JSONLoader
#method-1=JSONLOADER with JQ Schema
print("JSON Loader Extract Specifi fields")
employee_loader=JSONLoader(
    file_path='data\json-files\company_data.json',
    jq_schema='.employees[]',
    text_content=False    
)
employee_docs=employee_loader.load()
print(f'This is the length of the docs {len(employee_docs)}')
print(f'This  is the first docs{employee_docs[0]}')
```
