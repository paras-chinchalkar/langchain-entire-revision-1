```python
from langchain_core.documents import Document
docs=[
    Document(page_content="Langchain Helps in building LLM Applications"),
    Document(page_content="Pinecone is a vector database for semantic search"),
    Document(page_content="The Eiffel Tower is located in Paris"),
    Document(page_content="The Eiffel Tower is located in Paris"),
    Document(page_content="Langchain can be used to develop agentic AI Applications")
]
```

```python
#Embeddings
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(
    model='sentence-transformers/all-MiniLM-L6-v2'
)
```

```python
#Making the FIASS data vector
from langchain_community.vectorstores import FAISS
dense_vector=FAISS.from_documents(docs,embeddings)
```

```python
#Making an retriever
retriever=dense_vector.as_retriever()
```
