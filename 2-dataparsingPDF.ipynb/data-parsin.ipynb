```python
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyMuPDFLoader

class SmartPDFProcessor:
    """Advanced PDF Processing with error handling and smart chunking."""

    def __init__(self, chunk_size: int = 200, chunk_overlap: int = 20):
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
            separators=[" "]  # corrected spelling: 'separators'
        )

    def _clean_text(self, text: str) -> str:
        """Utility to clean and normalize text."""
        if not text:
            return ""
        # Example cleaning: strip whitespace, normalize spaces
        return " ".join(text.split())
        # #Fix common PDF Extraction issues 
        # text=text.replace("fi","fi")
        # text=text.replace("fi","fi")

    def process_pdf(self, pdf_path: str) -> list[Document]:
        """Process PDF with smart chunking and metadata enhancement."""
        try:
            loader = PyMuPDFLoader(pdf_path)
            pages = loader.load()
        except Exception as e:
            raise RuntimeError(f"Failed to load PDF {pdf_path}: {e}")

        processed_chunks = []
        for page_num, page in enumerate(pages):
            cleaned_text = self._clean_text(page.page_content)
            if len(cleaned_text.strip())<50:
                continue
            
            
            #Create Chunks with enhanced metadata 
            chunks=self.text_splitter.create_documents(
                texts=[cleaned_text],
                metadatas=[{
                    **page.metadata,
                    "page":page_num+1,
                    "total_pages":len(pages),
                    "chunk_method":"smart_pdf_processor",
                    "char_count":len(cleaned_text)
                }]
            )

        return processed_chunks
    
```

```python
#Method-1-using DocxtxtLoader
print("Using DocxtxtLoader")
try:
    docx_loader=Docx2txtLoader("2-dataparsingPDF.ipynb\Quotation Cable Laying 13.12.2025.docx")
    docs=docx_loader.load()
    print(f"Loaded {len(docs)} documents")
    print(f"Content preview {docs[0].page_content[:200]}...")
    print(f"MetaData {docs[0].metadata}")

except Exception as e:
    print("eRROR ")
```

```python
#Method2-unstrcutred docx reader
print("Using unstructuredWordDOcumentLoader")
try:
    unstructured=UnstructuredWordDocumentLoader("2-dataparsingPDF.ipynb\Quotation Cable Laying 13.12.2025.docx",mode='elements')
    loader=unstructured.load()
    print(f"the loaded element is {len(unstructured)}")
    for i,doc in enumerate(loader[:3]):
        print(f"Element {i+1}")
        print(f"Type {doc.metadata.get('category','unknown')}")
        print(f"Content{doc.page_content[:100]}...")

except Exception as e:
    print(e)
```
